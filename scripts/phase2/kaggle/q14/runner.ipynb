```python
# ---- Kaggle self-containment shim (auto-inserted by build_kaggle_queue.py) ----
import pathlib as _pl
_os_cwd = _pl.Path.cwd()
(_os_cwd / "data" / "phase2").mkdir(parents=True, exist_ok=True)
(_os_cwd / "scripts" / "phase2" / "out").mkdir(parents=True, exist_ok=True)
(_os_cwd / "data" / "phase2" / "q14_sleepfaces.csv").write_text('# Q14 (Phase 2): Does sleep deprivation hurt one\'s ability to interpret facial\n# expressions?\n#\n# STUDY\n#   Killgore WDS, Balkin TJ, Yarnell AM, Capaldi VF II. "Sleep deprivation\n#   impairs recognition of specific emotions." Neurobiology of Sleep and\n#   Circadian Rhythms 2017;3:10-16. doi:10.1016/j.nbscr.2017.01.001 (open\n#   access, CC BY-NC-ND; PMID 31236499, PMCID PMC6575584). NOTE: some\n#   secondary sources list a 2017 NeuroReport venue; the correct journal is\n#   Neurobiology of Sleep and Circadian Rhythms (verified against the\n#   published PDF, Oct 2026).\n#   ScienceDaily\'s coverage (21 Mar 2017) headlines this study as "Sleep\n#   deprivation impairs ability to interpret facial expressions", matching the\n#   Q14 question.\n#\n# DESIGN\n#   Within-subject experiment (Walter Reed Army Institute of Research). 54\n#   healthy young adults (29 male, 25 female; mean age 23.5, SD 4.0) took the\n#   Ekman Hexagon Test (EHT): 120 computer-morphed facial blends (20 per basic\n#   emotion; happiness, sadness, surprise, fear, disgust, anger; 50:50 blends\n#   excluded) classified under the 6 emotion labels. Tested after a normal\n#   night of sleep (baseline, ~5.5 h awake), after one night of total sleep\n#   deprivation (23.5 h awake; a 46.7-h post-stimulant session reported\n#   elsewhere is excluded), and after 12 h recovery sleep. Within-subject:\n#   every participant completed every session.\n#\n# PUBLISHED RESULTS (repeated-measures ANOVA + planned contrasts)\n#   Session x emotion interaction F(10,43)=3.69, p=.001.\n#   Happiness: decline baseline->SD p=.004 (quadratic F(1,52)=7.12, p=.01)\n#   Sadness:   decline baseline->SD p=.023 (quadratic F(1,52)=5.85, p=.019)\n#   Surprise:  no decline (p=.39); Fear: no baseline->SD decline (p=.39)\n#   Disgust / Anger: no significant effects (quadratic p=.33 / p=.41)\n#   Recovery sleep returned accuracy to baseline. Numeric per-condition means\n#   appear only in Fig. 3 (bar-free line plots); no table is published.\n#\n# DATA SOURCE\n#   Lock5 distributes no dataset for this study (3rd-edition data page checked\n#   Oct 2026 - only SleepCaffeine and SleepStudy exist, neither is this\n#   experiment). data/phase2/q14_sleepfaces.csv is a documented\n#   reconstruction, for method illustration only (true per-participant values\n#   are unpublished).\n#\n# DIGITIZED MEANS (percent correct; read programmatically off Fig. 3 at 600 dpi\n#   by calibrating the y-axis tick-label text positions and locating the plotted\n#   marker centroids; reading tolerance ~+/-0.15 point):\n#                 rested(baseline)  sleep-dep   recovery(not included)\n#   Happiness          99.91          98.63        99.75\n#   Sadness            98.75          96.09        98.75\n#   Surprise           88.46          87.96        92.98\n#   Fear               91.62          90.18        96.38\n#   Disgust            84.58          87.77        87.21\n#   Anger              83.56          85.34        91.06\n#   (happiness rested: digitized 99.95 is unattainable on the 5-point grid of\n#   correct/20 trials; 99.91 = one person at 95, everyone else 100, is the\n#   nearest attainable value and matches the paper\'s stated ceiling effect.)\n#\n# RECONSTRUCTION\n#   - Unit: one row per participant (n = 54). Each cell = that participant\'s\n#     percent correct (out of 20 morphed faces per emotion; hence 5-point\n#     steps) for the rested baseline and the sleep-deprived session.\n#   - Condition means: exactly the digitized values above (integer error-count\n#     adjustment after generation).\n#   - Paired-difference SDs calibrated so the paired t (df=53) approximately\n#     reproduces the published planned-contrast p: happiness .004 -> SD 3.1;\n#     sadness .023 -> 8.4; surprise .39 -> (model floor) 8.9; fear .39 -> 12.2;\n#     disgust/anger (no simple-effect p published) -> 14.0/8.0 so their small\n#     improvements remain non-significant, consistent with the paper.\n#   - Generation model (seed 2026, numpy default_rng): per person and emotion,\n#     errors ~ Binomial(20, q) with rested q = expit(a + g*z) and deprived\n#     q = expit(a + D + g*z + l*u); z = face-recognition ability factor\n#     (cross-emotion corr 0.5), u = sleep-loss vulnerability factor (corr 0.3),\n#     independent person factors, rank-matched to N(0,1) quantiles. Calibrated\n#     parameters (a, g, D, l) per emotion:\n#       happiness (-7.013, 0.050, 2.245, 1.018); sadness (-8.226, 3.203, 1.725, 0.417)\n#       surprise  (-2.927, 1.707, 0.064, 0.000); fear    (-3.028, 1.319, -0.065, 0.960)\n#       disgust   (-3.356, 2.783, -0.702, 1.130); anger  (-3.043, 2.573, -0.235, 0.000)\n#   - Realized paired results of this CSV: happiness t(53)=-2.81, p=.007;\n#     sadness t=-2.75, p=.008; surprise/fear/disgust/anger n.s. (p=.72/.52/.07/.09);\n#     overall (mean of 6 emotions) t=-0.18, p=.86 - averaging across emotions\n#     masks the emotion-specific effect.\n#   - Recovery-session data are not included (analysis contrasts rested vs\n#     sleep-deprived only).\nparticipant,happiness_rested,happiness_deprived,sadness_rested,sadness_deprived,surprise_rested,surprise_deprived,fear_rested,fear_deprived,disgust_rested,disgust_deprived,anger_rested,anger_deprived\n1,100.0,100.0,100.0,100.0,100.0,100.0,95.0,90.0,100.0,95.0,100.0,100.0\n2,100.0,100.0,100.0,100.0,95.0,95.0,95.0,95.0,100.0,100.0,70.0,80.0\n3,100.0,95.0,95.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0\n4,100.0,100.0,100.0,100.0,50.0,50.0,90.0,90.0,85.0,100.0,85.0,100.0\n5,100.0,100.0,100.0,95.0,90.0,60.0,100.0,100.0,95.0,95.0,85.0,90.0\n6,100.0,100.0,100.0,95.0,100.0,100.0,100.0,85.0,100.0,100.0,100.0,90.0\n7,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,70.0,75.0\n8,100.0,100.0,100.0,100.0,100.0,100.0,95.0,95.0,100.0,100.0,100.0,100.0\n9,100.0,95.0,95.0,100.0,80.0,95.0,100.0,95.0,100.0,100.0,100.0,95.0\n10,100.0,100.0,100.0,95.0,100.0,100.0,95.0,100.0,100.0,100.0,100.0,100.0\n11,100.0,100.0,95.0,95.0,100.0,95.0,95.0,95.0,45.0,95.0,90.0,95.0\n12,100.0,100.0,100.0,100.0,95.0,100.0,90.0,90.0,90.0,95.0,65.0,50.0\n13,100.0,100.0,95.0,100.0,100.0,90.0,100.0,95.0,100.0,100.0,100.0,100.0\n14,100.0,100.0,100.0,100.0,95.0,100.0,75.0,95.0,100.0,100.0,80.0,85.0\n15,100.0,100.0,100.0,100.0,65.0,70.0,95.0,100.0,100.0,100.0,100.0,95.0\n16,100.0,95.0,100.0,100.0,95.0,95.0,95.0,85.0,100.0,100.0,100.0,100.0\n17,100.0,100.0,100.0,100.0,95.0,100.0,100.0,95.0,100.0,100.0,100.0,100.0\n18,100.0,100.0,100.0,100.0,95.0,90.0,80.0,80.0,85.0,90.0,10.0,15.0\n19,100.0,100.0,100.0,100.0,50.0,25.0,90.0,100.0,95.0,95.0,100.0,95.0\n20,100.0,100.0,95.0,100.0,100.0,100.0,100.0,95.0,100.0,100.0,100.0,100.0\n21,95.0,100.0,100.0,100.0,95.0,95.0,95.0,95.0,100.0,100.0,90.0,90.0\n22,100.0,100.0,100.0,95.0,75.0,60.0,85.0,95.0,65.0,100.0,75.0,70.0\n23,100.0,95.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0\n24,100.0,100.0,100.0,95.0,80.0,95.0,100.0,100.0,100.0,100.0,100.0,100.0\n25,100.0,100.0,100.0,90.0,90.0,100.0,90.0,90.0,90.0,85.0,100.0,95.0\n26,100.0,100.0,95.0,80.0,90.0,95.0,70.0,55.0,20.0,40.0,60.0,50.0\n27,100.0,95.0,100.0,100.0,95.0,95.0,100.0,100.0,100.0,90.0,100.0,100.0\n28,100.0,100.0,100.0,100.0,100.0,100.0,100.0,95.0,100.0,95.0,95.0,100.0\n29,100.0,100.0,100.0,100.0,100.0,100.0,95.0,100.0,100.0,100.0,100.0,100.0\n30,100.0,100.0,100.0,100.0,90.0,100.0,75.0,60.0,60.0,65.0,70.0,95.0\n31,100.0,100.0,100.0,100.0,95.0,95.0,100.0,95.0,100.0,95.0,85.0,95.0\n32,100.0,100.0,100.0,95.0,85.0,55.0,90.0,100.0,100.0,100.0,50.0,45.0\n33,100.0,100.0,100.0,100.0,100.0,100.0,85.0,100.0,55.0,80.0,95.0,100.0\n34,100.0,100.0,100.0,100.0,80.0,85.0,95.0,95.0,25.0,60.0,10.0,25.0\n35,100.0,95.0,100.0,100.0,100.0,100.0,100.0,95.0,100.0,100.0,95.0,100.0\n36,100.0,100.0,100.0,100.0,95.0,100.0,100.0,90.0,90.0,95.0,95.0,90.0\n37,100.0,100.0,100.0,95.0,100.0,100.0,95.0,100.0,100.0,95.0,100.0,100.0\n38,100.0,100.0,100.0,95.0,100.0,100.0,95.0,95.0,90.0,100.0,100.0,100.0\n39,100.0,100.0,100.0,100.0,30.0,10.0,85.0,60.0,95.0,100.0,100.0,100.0\n40,100.0,100.0,100.0,100.0,80.0,85.0,95.0,100.0,50.0,50.0,95.0,100.0\n41,100.0,100.0,100.0,100.0,95.0,100.0,40.0,85.0,85.0,80.0,100.0,95.0\n42,100.0,100.0,100.0,100.0,95.0,95.0,90.0,90.0,100.0,100.0,75.0,90.0\n43,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0,95.0\n44,100.0,100.0,100.0,100.0,70.0,85.0,95.0,80.0,100.0,100.0,90.0,80.0\n45,100.0,100.0,100.0,100.0,90.0,90.0,95.0,70.0,90.0,95.0,100.0,95.0\n46,100.0,100.0,100.0,100.0,100.0,95.0,100.0,90.0,95.0,100.0,35.0,55.0\n47,100.0,95.0,100.0,90.0,35.0,45.0,90.0,85.0,90.0,90.0,60.0,60.0\n48,100.0,100.0,100.0,100.0,95.0,100.0,100.0,100.0,100.0,100.0,100.0,100.0\n49,100.0,100.0,95.0,65.0,60.0,75.0,85.0,85.0,25.0,10.0,35.0,50.0\n50,100.0,100.0,70.0,45.0,75.0,70.0,30.0,90.0,5.0,40.0,0.0,5.0\n51,100.0,95.0,95.0,90.0,95.0,95.0,95.0,60.0,30.0,30.0,100.0,100.0\n52,100.0,85.0,100.0,100.0,95.0,90.0,95.0,45.0,30.0,0.0,100.0,100.0\n53,100.0,95.0,100.0,100.0,100.0,100.0,95.0,95.0,100.0,100.0,100.0,100.0\n54,100.0,85.0,100.0,75.0,85.0,75.0,95.0,95.0,80.0,80.0,45.0,65.0\n')
import os as _os
_os.chdir(_os_cwd)
REPO = _os_cwd  # overridden for Kaggle (original assignment below is ignored)
# ---- end shim ----
"""Q14 (Phase 2): Does sleep deprivation hurt one's ability to interpret
facial expressions?

Provenance
----------
Killgore WDS, Balkin TJ, Yarnell AM, Capaldi VF II. "Sleep deprivation
impairs recognition of specific emotions." Neurobiology of Sleep and
Circadian Rhythms 2017;3:10-16. doi:10.1016/j.nbscr.2017.01.001 (open
access, CC BY-NC-ND; PMID 31236499, PMCID PMC6575584). Some secondary
sources list a 2017 NeuroReport venue; the correct journal is Neurobiology
of Sleep and Circadian Rhythms (verified against the published PDF, Oct
2026). ScienceDaily (21 Mar 2017) headlined this study as "Sleep
deprivation impairs ability to interpret facial expressions", matching the
Q14 question.

Design: within-subject experiment (Walter Reed Army Institute of Research).
54 healthy young adults (29 male, 25 female; mean age 23.5, SD 4.0) took
the Ekman Hexagon Test (EHT): 120 computer-morphed facial blends (20 per
basic emotion - happiness, sadness, surprise, fear, disgust, anger; 50:50
blends excluded) classified under the 6 emotion labels. Tested after a
normal night of sleep (baseline), after one night of total sleep
deprivation (23.5 h awake), and after 12 h recovery sleep. Published
results: happiness and sadness recognition declined from baseline to sleep
deprivation (planned contrasts p=.004 and p=.023); surprise, fear, disgust
and anger were unaffected; recovery sleep returned accuracy to baseline.
Numeric per-condition means appear only in Fig. 3 (no table is published).

Data: Lock5 distributes no dataset for this study (3rd-edition data page
checked Oct 2026 - only SleepCaffeine and SleepStudy exist, neither is this
experiment). data/phase2/q14_sleepfaces.csv is a documented reconstruction:
54 participant-level rows, percent correct (out of 20 trials per emotion,
5-point steps) per emotion for the rested and sleep-deprived sessions.
Condition means exactly match Fig. 3 values read programmatically off the
published figure (axis-tick text calibration + marker centroid detection,
600 dpi, tolerance ~+/-0.15 point); paired-difference SDs are calibrated so
the paired t (df=53) approximately reproduces the published planned-contrast
p-values. Generation algorithm, calibrated parameters and seed (2026) are
documented in the CSV header. True per-participant values are unpublished -
use for method illustration, not re-estimation.

Analysis (matched pairs, Lock5 spirit)
--------------------------------------
  - Paired t-test on the 54 within-participant differences
    (deprived - rested), overall and per emotion
  - 100,000-rep sign-flip randomization test on the differences
  - 100,000-rep bootstrap percentile CI of the mean difference
    (resampling participants, i.e. whole pairs)

Run from repo root: .venv/bin/python scripts/phase2/q14_sleepfaces.py
Outputs: prints summary; writes scripts/phase2/out/q14_results.json
"""
import json
from pathlib import Path

import numpy as np
import pandas as pd
from scipy import stats

REPO = _os_cwd  # Kaggle override
CSV_PATH = REPO / "data" / "phase2" / "q14_sleepfaces.csv"
OUT_PATH = REPO / "scripts" / "phase2" / "out" / "q14_results.json"

N_REPS = 100_000
SEED = 2026
rng = np.random.default_rng(SEED)

EMOTIONS = ["happiness", "sadness", "surprise", "fear", "disgust", "anger"]

# digitized from Fig. 3 of the paper (percent correct)
PUB_MEANS = {
    "happiness": (99.91, 98.63), "sadness": (98.75, 96.09),
    "surprise": (88.46, 87.96),  "fear": (91.62, 90.18),
    "disgust": (84.58, 87.77),   "anger": (83.56, 85.34),
}
# published planned contrasts (baseline vs sleep deprivation)
PUB_P = {"happiness": .004, "sadness": .023, "surprise": .39,
         "fear": .39, "disgust": None, "anger": None}

# ---------------------------------------------------------------- load data
df = pd.read_csv(CSV_PATH, comment="#")
n = len(df)
assert n == 54, f"expected 54 participants, got {n}"

results = {
    "question": "Does sleep deprivation hurt one's ability to interpret "
                "facial expressions?",
    "provenance": {
        "citation": "Killgore WDS, Balkin TJ, Yarnell AM, Capaldi VF II. "
                    "Sleep deprivation impairs recognition of specific "
                    "emotions. Neurobiology of Sleep and Circadian Rhythms "
                    "2017;3:10-16. doi:10.1016/j.nbscr.2017.01.001 (PMID "
                    "31236499, PMCID PMC6575584).",
        "citation_note": "Some secondary sources list a 2017 NeuroReport "
                         "venue; the correct journal is Neurobiology of Sleep "
                         "and Circadian Rhythms (verified against the "
                         "published PDF, Oct 2026).",
        "design": "Within-subject experiment: 54 healthy young adults (29 M, "
                  "25 F) classified 120 morphed facial expressions (20 per "
                  "basic emotion) on the Ekman Hexagon Test after a normal "
                  "night of sleep, after one night of total sleep deprivation "
                  "(23.5 h awake), and after 12 h recovery sleep",
        "data_source": "Reconstructed CSV (data/phase2/q14_sleepfaces.csv): "
                       "condition means digitized programmatically from Fig. 3 "
                       "of the paper; paired-difference SDs calibrated to the "
                       "published planned-contrast p-values. No Lock5 dataset "
                       "exists for this study (3e data page checked Oct 2026). "
                       "Generation algorithm documented in the CSV header.",
    },
    "data": {"n_participants": n, "n_trials_per_emotion": 20,
             "scale": "percent correct out of 20 morphed faces (5-point steps)",
             "source_csv": str(CSV_PATH.relative_to(REPO))},
    "methods": {"paired_test": "scipy.stats.ttest_rel (paired t, df = 53)",
                "randomization": "sign-flip test on differences, two-sided",
                "bootstrap": "percentile CI of mean difference, resampling "
                             "participants (whole pairs)",
                "n_replicates": N_REPS, "seed": SEED,
                "engine": "numpy/scipy/pandas"},
    "outcomes": {},
}


def analyze(rested, deprived, label, pub):
    diff = deprived - rested  # <0 means worse recognition when deprived
    mean_diff = float(diff.mean())
    sd_diff = float(diff.std(ddof=1))

    t_res = stats.ttest_rel(deprived, rested)

    # sign-flip randomization test: under H0 differences are symmetric about 0
    signs = rng.choice([-1.0, 1.0], size=(N_REPS, n))
    flip_means = (signs * diff).mean(axis=1)
    p_flip = float((np.sum(np.abs(flip_means) >= abs(mean_diff)) + 1)
                   / (N_REPS + 1))

    # bootstrap percentile CI of the mean difference (resample participants)
    idx = rng.integers(0, n, size=(N_REPS, n))
    boot_means = diff[idx].mean(axis=1)
    ci_boot = np.percentile(boot_means, [2.5, 97.5])

    return {
        "published": {"mean_rested": pub[0], "mean_deprived": pub[1],
                      "p_paper": PUB_P.get(label)},
        "reconstructed": {"mean_rested": float(rested.mean()),
                          "mean_deprived": float(deprived.mean()),
                          "sd_rested": float(rested.std(ddof=1)),
                          "sd_deprived": float(deprived.std(ddof=1))},
        "mean_difference_deprived_minus_rested": mean_diff,
        "sd_difference": sd_diff,
        "paired_t": {"t": float(t_res.statistic), "df": n - 1,
                     "p_two_sided": float(t_res.pvalue)},
        "p_signflip_randomization": p_flip,
        "ci_95_bootstrap_percentile": [float(ci_boot[0]), float(ci_boot[1])],
    }


# ------------------------------------------------- matched-pairs analysis
for label in EMOTIONS:
    rested = df[f"{label}_rested"].to_numpy(dtype=float)
    deprived = df[f"{label}_deprived"].to_numpy(dtype=float)
    results["outcomes"][label] = analyze(rested, deprived, label,
                                         PUB_MEANS[label])

# overall: participant mean accuracy across the six emotions
rested = df[[f"{e}_rested" for e in EMOTIONS]].mean(axis=1).to_numpy()
deprived = df[[f"{e}_deprived" for e in EMOTIONS]].mean(axis=1).to_numpy()
pub_overall = (float(np.mean([PUB_MEANS[e][0] for e in EMOTIONS])),
               float(np.mean([PUB_MEANS[e][1] for e in EMOTIONS])))
results["outcomes"]["overall"] = analyze(rested, deprived, "overall",
                                         pub_overall)

OUT_PATH.parent.mkdir(parents=True, exist_ok=True)
with open(OUT_PATH, "w") as f:
    json.dump(results, f, indent=2)

# ------------------------------------------------- human-readable summary
print("=" * 64)
print("Q14: Does sleep deprivation hurt one's ability to interpret")
print("facial expressions?")
print("Killgore et al., Neurobiol Sleep Circadian Rhythms 2017;3:10-16 -")
print("54 adults, Ekman Hexagon Test, rested vs sleep-deprived (paired);")
print("reconstructed participant-level data (means digitized from Fig. 3)")
print("=" * 64)
for label in EMOTIONS + ["overall"]:
    o = results["outcomes"][label]
    md = o["mean_difference_deprived_minus_rested"]
    ci = o["ci_95_bootstrap_percentile"]
    hdr = "Overall (mean of 6 emotions)" if label == "overall" \
        else label.capitalize()
    print(f"\n{hdr}:")
    print(f"  rested = {o['reconstructed']['mean_rested']:.2f}%, "
          f"sleep-deprived = {o['reconstructed']['mean_deprived']:.2f}% "
          f"(digitized from Fig. 3: {o['published']['mean_rested']:.2f} vs "
          f"{o['published']['mean_deprived']:.2f})")
    print(f"  mean difference (deprived - rested) = {md:+.2f} points")
    print(f"  paired t({o['paired_t']['df']}) = {o['paired_t']['t']:.2f}, "
          f"p = {o['paired_t']['p_two_sided']:.4f}")
    print(f"  sign-flip randomization p = {o['p_signflip_randomization']:.4f}")
    print(f"  bootstrap 95% CI of mean difference: ({ci[0]:+.2f}, {ci[1]:+.2f})")

sig = [e for e in EMOTIONS
       if results["outcomes"][e]["paired_t"]["p_two_sided"] < 0.05]
print("\nConclusion: sleep deprivation significantly impaired recognition of"
      f" {', '.join(sig)} - the prosocial emotions - while recognition of")
print("surprise, fear, disgust and anger (threat-relevant expressions) was")
print("unaffected. Overall accuracy across all six emotions was unchanged,")
print("so averaging across emotions masks the emotion-specific deficit;")
print("accuracy returned to baseline after recovery sleep (per the paper).")
print(f"\nResults JSON written to {OUT_PATH.relative_to(REPO)}")
```
